```python
import pandas as pd
import numpy as np

# Load the merged daily data
df = pd.read_csv("data/processed/factors_daily.csv", index_col=0, parse_dates=True)

print(f"Shape: {df.shape}")
print(f"Date range: {df.index.min().date()} to {df.index.max().date()}")
print(f"\nColumns: {list(df.columns)}")

# Check btc_price integrity — this is the critical column
print(f"\n--- btc_price checks ---")
print(f"  Zeros: {(df['btc_price'] == 0).sum()}")
print(f"  NaNs: {df['btc_price'].isna().sum()}")
print(f"  Negatives: {(df['btc_price'] < 0).sum()}")
print(f"  Min: {df['btc_price'].min()}")
print(f"  Max: {df['btc_price'].max()}")

# Check for date gaps (missing days)
full_range = pd.date_range(df.index.min(), df.index.max(), freq="D")
missing_days = len(full_range) - len(df.index.unique())
print(f"\n  Calendar days in range: {len(full_range)}")
print(f"  Actual rows: {len(df)}")
print(f"  Missing/duplicate days: {missing_days}")
```

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt
%matplotlib inline

GENESIS = pd.Timestamp("2009-01-03")
TRAIN_END = "2022-12-31"

# Days since genesis and logs
df["days_since_genesis"] = (df.index - GENESIS).days
df["log_price"] = np.log(df["btc_price"])
df["log_days"] = np.log(df["days_since_genesis"])

# Fit power-law trend on TRAINING DATA ONLY (leakage-free)
train = df.loc[:TRAIN_END]
X_train = sm.add_constant(train["log_days"])
y_train = train["log_price"]
trend_model = sm.OLS(y_train, X_train).fit()

alpha, beta = trend_model.params
print(f"Reference power-law trend (fit on 2015-2022 only):")
print(f"  alpha = {alpha:.4f}, beta = {beta:.4f}, R² = {trend_model.rsquared:.4f}")

# Project trend across ALL dates and compute deviation
df["trend"] = alpha + beta * df["log_days"]
df["deviation"] = df["log_price"] - df["trend"]

# Split deviation
dev_train = df.loc[:TRAIN_END, "deviation"]
dev_test = df.loc["2023-01-01":, "deviation"]

print(f"\nDeviation series:")
print(f"  Train ({len(dev_train)} obs): mean={dev_train.mean():.3f}, std={dev_train.std():.3f}")
print(f"  Test  ({len(dev_test)} obs): mean={dev_test.mean():.3f}, std={dev_test.std():.3f}")
print(f"  Full range: min={df['deviation'].min():.3f}, max={df['deviation'].max():.3f}")

# Sanity plot
fig, axes = plt.subplots(2, 1, figsize=(13, 7))
axes[0].plot(df.index, df["log_price"], label="log price", linewidth=0.8)
axes[0].plot(df.index, df["trend"], label="reference trend", color="red", linewidth=1.5)
axes[0].axvline(pd.Timestamp(TRAIN_END), color="gray", linestyle=":", label="train/test split")
axes[0].legend(); axes[0].set_title("Log price vs reference trend (fit on train only)")
axes[1].plot(df.index, df["deviation"], color="green", linewidth=0.8)
axes[1].axhline(0, color="black", linewidth=0.5)
axes[1].axvline(pd.Timestamp(TRAIN_END), color="gray", linestyle=":")
axes[1].set_title("Deviation (SETAR target)")
plt.tight_layout(); plt.show()
```

```python
# Load the FULL price history (independent of factor start date)
price = pd.read_csv("data/raw/btc_price.csv", index_col="date", parse_dates=True)

TREND_START = "2013-01-01"
TRAIN_END = "2022-12-31"

# Fit trend on 2013+ training data only
price_fit = price.loc[TREND_START:TRAIN_END].copy()
price_fit["days"] = (price_fit.index - GENESIS).days
price_fit["log_price"] = np.log(price_fit["btc_price"])
price_fit["log_days"] = np.log(price_fit["days"])

X = sm.add_constant(price_fit["log_days"])
trend_model = sm.OLS(price_fit["log_price"], X).fit()
alpha, beta = trend_model.params
print(f"Trend fit on {TREND_START} to {TRAIN_END}:")
print(f"  alpha={alpha:.4f}, beta={beta:.4f}, R²={trend_model.rsquared:.4f}")

# Now build the deviation on the FACTOR window (2015+, where df lives)
# so it aligns with everything else, but using the better-fit trend
df["days_since_genesis"] = (df.index - GENESIS).days
df["log_price"] = np.log(df["btc_price"])
df["log_days"] = np.log(df["days_since_genesis"])
df["trend"] = alpha + beta * df["log_days"]
df["deviation"] = df["log_price"] - df["trend"]

dev_train = df.loc[:TRAIN_END, "deviation"]
dev_test = df.loc["2023-01-01":, "deviation"]
print(f"\nDeviation (on 2015+ factor window, using 2013-fit trend):")
print(f"  Train: mean={dev_train.mean():.3f}, std={dev_train.std():.3f}")
print(f"  Test:  mean={dev_test.mean():.3f}, std={dev_test.std():.3f}")
print(f"  Range: [{df['deviation'].min():.3f}, {df['deviation'].max():.3f}]")
```

```python
from sklearn.metrics import mean_squared_error

# Prepare deviation series aligned with the factor window
dev = df["deviation"].dropna()
TRAIN_END = "2022-12-31"

dev_train = dev.loc[:TRAIN_END]
dev_test = dev.loc["2023-01-01":]

# Build lag-1 series (SETAR needs y_t and y_{t-1})
def build_lag_frame(series):
    lag_df = pd.DataFrame({
        "y": series,
        "y_lag1": series.shift(1)
    }).dropna()
    return lag_df

lag_train = build_lag_frame(dev_train)
lag_test = build_lag_frame(dev_test)

# --- Threshold search ---
# Grid over inner 60% of training-period deviation values
low_q, high_q = dev_train.quantile(0.20), dev_train.quantile(0.80)
threshold_grid = np.linspace(low_q, high_q, 100)

def fit_setar_at_threshold(lag_df, threshold):
    """Fit 2-regime AR(1) SETAR at a given threshold. Returns coeffs + SSR."""
    below = lag_df[lag_df["y_lag1"] <= threshold]
    above = lag_df[lag_df["y_lag1"] > threshold]
    if len(below) < 30 or len(above) < 30:
        return None
    # Regime 1: y = a1 + b1 * y_lag1
    X1 = sm.add_constant(below["y_lag1"])
    m1 = sm.OLS(below["y"], X1).fit()
    # Regime 2
    X2 = sm.add_constant(above["y_lag1"])
    m2 = sm.OLS(above["y"], X2).fit()
    ssr = m1.ssr + m2.ssr
    return {"threshold": threshold, "m1": m1, "m2": m2, "ssr": ssr,
            "n_below": len(below), "n_above": len(above)}

# Search
results = []
for t in threshold_grid:
    fit = fit_setar_at_threshold(lag_train, t)
    if fit:
        results.append(fit)

best = min(results, key=lambda r: r["ssr"])
print(f"Optimal threshold: {best['threshold']:.4f}")
print(f"  Below-regime obs: {best['n_below']} | Above-regime obs: {best['n_above']}")
print(f"  Regime 1 (below):  a={best['m1'].params[0]:.4f}, b={best['m1'].params[1]:.4f}")
print(f"  Regime 2 (above):  a={best['m2'].params[0]:.4f}, b={best['m2'].params[1]:.4f}")

# --- Out-of-sample prediction ---
def predict_setar(lag_df, best):
    """Apply the fitted SETAR to new data."""
    y_lag = lag_df["y_lag1"].values
    a1, b1 = best["m1"].params
    a2, b2 = best["m2"].params
    below_mask = y_lag <= best["threshold"]
    preds = np.where(below_mask, a1 + b1 * y_lag, a2 + b2 * y_lag)
    return preds

train_preds = predict_setar(lag_train, best)
test_preds = predict_setar(lag_test, best)

train_rmse = np.sqrt(mean_squared_error(lag_train["y"], train_preds))
test_rmse = np.sqrt(mean_squared_error(lag_test["y"], test_preds))

# Baseline: predict deviation = 0 (the "power law only" model)
baseline_test_rmse = np.sqrt(mean_squared_error(lag_test["y"], np.zeros_like(lag_test["y"])))

# Baseline 2: naive AR(1) with no threshold (single regime)
X_naive = sm.add_constant(lag_train["y_lag1"])
m_naive = sm.OLS(lag_train["y"], X_naive).fit()
naive_test_preds = m_naive.params[0] + m_naive.params[1] * lag_test["y_lag1"].values
naive_test_rmse = np.sqrt(mean_squared_error(lag_test["y"], naive_test_preds))

print(f"\n=== Results ===")
print(f"SETAR train RMSE:      {train_rmse:.4f}")
print(f"SETAR test RMSE:       {test_rmse:.4f}")
print(f"Naive AR(1) test RMSE: {naive_test_rmse:.4f}")
print(f"Power-law test RMSE:   {baseline_test_rmse:.4f}")
```

```python
import pandas as pd, numpy as np
from pysr import PySRRegressor

# Load data
df = pd.read_csv("data/processed/factors_daily.csv", index_col=0, parse_dates=True)

# Input: non-zero-balance addresses (Metcalfe network size). Target: log price.
d = df[["btc_price", "addresses_nonzero_balance"]].dropna().copy()
d["log_price"] = np.log(d["btc_price"])
d["log_addr"] = np.log(d["addresses_nonzero_balance"])

# Cutoff: fit pre-2026, test on 2026 (out-of-sample)
CUTOFF = "2026-01-01"
train = d.loc[:CUTOFF]
test = d.loc[CUTOFF:]

X_train = train[["log_addr"]].values
y_train = train["log_price"].values

print(f"Train: {len(train)} ({train.index.min().date()} to {train.index.max().date()})")
print(f"Test:  {len(test)} ({test.index.min().date()} to {test.index.max().date()})")

model = PySRRegressor(
    niterations=100,
    binary_operators=["+", "-", "*", "/"],
    unary_operators=["log", "sqrt"],
    nested_constraints={"log": {"log": 0, "sqrt": 1}, "sqrt": {"sqrt": 0, "log": 1}},
    maxsize=10,
    elementwise_loss="loss(pred, target) = (pred - target)^2",
    model_selection="best",
    random_state=0,
    deterministic=True,
    parallelism="serial",
)

model.fit(X_train, y_train)

print("\n=== Pareto front ===")
print(model.equations_[["complexity", "loss", "equation"]])

pred_test = model.predict(test[["log_addr"]].values)
test_rmse = np.sqrt(np.mean((test["log_price"].values - pred_test)**2))
print(f"\nBest equation: {model.sympy()}")
print(f"Out-of-sample (2026) RMSE: {test_rmse:.4f}")
```

```python
import pandas as pd, numpy as np
from pysr import PySRRegressor

df = pd.read_csv("data/processed/factors_daily.csv", index_col=0, parse_dates=True)

# Inputs: addresses (log), etf_flows (raw), global_liquidity (log)
d = df[["btc_price", "addresses_nonzero_balance", "etf_flows_usd", "global_liquidity_t"]].copy()
d["log_price"] = np.log(d["btc_price"])
d["log_addr"]  = np.log(d["addresses_nonzero_balance"])
d["flows"]     = d["etf_flows_usd"].fillna(0) / 1e9          # scale to $B, 0 pre-ETF
d["log_liq"]   = np.log(d["global_liquidity_t"].replace(0, np.nan))
d = d.dropna(subset=["log_price", "log_addr", "log_liq"])

INPUTS = ["log_addr", "flows", "log_liq"]

regimes = [
    ("R1_early",      None,         "2016-07-09"),
    ("R2_2016-2020",  "2016-07-09", "2020-05-11"),
    ("R3_2020-2024",  "2020-05-11", "2024-01-11"),
    ("R4_ETF_era",    "2024-01-11", None),
]

def fit_regime(sub, name):
    if len(sub) < 200:
        print(f"  {name}: SKIP ({len(sub)} obs)"); return
    # within-regime 80/20 chronological split
    k = int(len(sub)*0.8)
    tr, te = sub.iloc[:k], sub.iloc[k:]
    m = PySRRegressor(
        niterations=60,
        binary_operators=["+", "-", "*", "/"],
        unary_operators=["log", "sqrt"],
        nested_constraints={"log":{"log":0,"sqrt":1},"sqrt":{"sqrt":0,"log":1}},
        maxsize=8,
        elementwise_loss="loss(pred, target) = (pred - target)^2",
        model_selection="best",
        random_state=0, deterministic=True, parallelism="serial",
        verbosity=0, progress=False,
    )
    m.fit(tr[INPUTS].values, tr["log_price"].values)
    tr_rmse = np.sqrt(np.mean((tr["log_price"].values - m.predict(tr[INPUTS].values))**2))
    te_rmse = np.sqrt(np.mean((te["log_price"].values - m.predict(te[INPUTS].values))**2))
    eqn = str(m.sympy())
    # which inputs are used?
    used = [INPUTS[i] for i in range(3) if f"x{i}" in eqn]
    flag = "  <-- OVERFIT?" if te_rmse > 2.5*tr_rmse else ""
    print(f"  {name} (n={len(sub)}): train_rmse={tr_rmse:.3f}  test_rmse={te_rmse:.3f}{flag}")
    print(f"      uses {used}")
    print(f"      {eqn}\n")

print("Per-regime SR (menu: addresses, etf_flows, liquidity) with within-regime hold-out\n")
for name, s, e in regimes:
    sub = d.loc[s:e] if (s or e) else d
    if s and e: sub = d.loc[s:e]
    elif s: sub = d.loc[s:]
    elif e: sub = d.loc[:e]
    fit_regime(sub, name)
```

```python
import pandas as pd, numpy as np
import statsmodels.api as sm

df = pd.read_csv("data/processed/factors_daily.csv", index_col=0, parse_dates=True)
d = df[["btc_price", "addresses_nonzero_balance"]].dropna().copy()
d["log_price"] = np.log(d["btc_price"])
d["log_addr"] = np.log(d["addresses_nonzero_balance"])

CUTOFF = "2024-01-11"   # train pre-ETF, test on ETF era (true out-of-sample)

def test_horizon(h):
    # past h-day address change -> next h-day price change
    X = d["log_addr"].diff(h).shift(h)      # address growth in the PRIOR window
    y = d["log_price"].diff(h)              # price growth in the CURRENT window
    data = pd.DataFrame({"x": X, "y": y}).dropna()

    tr = data.loc[:CUTOFF]; te = data.loc[CUTOFF:]

    # Newey-West regression (corrects for overlapping-window autocorrelation)
    Xtr = sm.add_constant(tr["x"])
    m = sm.OLS(tr["y"], Xtr).fit(cov_type="HAC", cov_kwds={"maxlags": h})
    beta = m.params.iloc[1]; pval = m.pvalues.iloc[1]; r2 = m.rsquared

    # out-of-sample: use train coeffs to predict test, compare to naive (predict 0)
    pred = m.params.iloc[0] + m.params.iloc[1]*te["x"]
    ss_res = ((te["y"]-pred)**2).sum()
    ss_naive = (te["y"]**2).sum()
    oos_r2 = 1 - ss_res/ss_naive   # >0 means beats predicting no-change

    print(f"  {h}-day horizon:")
    print(f"    train: beta={beta:.3f}, p={pval:.4f} (Newey-West), R²={r2:.3f}")
    print(f"    out-of-sample R² (ETF era): {oos_r2:.3f}  {'(beats naive)' if oos_r2>0 else '(worse than naive)'}")

print("Rigorous lagged address->price test (overlap-corrected + out-of-sample)\n")
for h in [30, 90]:
    test_horizon(h)
```

```python
import pandas as pd, numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

price = pd.read_csv("data/raw/btc_price.csv", index_col="date", parse_dates=True)
d = price.copy()
d["days"] = (d.index - pd.Timestamp("2009-01-03")).days
d["log_price"] = np.log(d["btc_price"])
d["log_days"] = np.log(d["days"])
d = d[d.index >= "2013-01-01"]

CUTOFF = "2026-01-01"
tr = d.loc[:CUTOFF]; te = d.loc[CUTOFF:]

# --- Plain power law: log_price = a + b*log_days ---
b_pl, a_pl = np.polyfit(tr["log_days"], tr["log_price"], 1)
pl_pred_te = a_pl + b_pl*te["log_days"]
pl_rmse = np.sqrt(np.mean((te["log_price"]-pl_pred_te)**2))

# --- Stretched exponential: log_price = A - B * days^(-C) ---
# As C small, curve bends (decelerates). Fit on train only.
def stretched(days, A, B, C):
    return A - B * np.power(days, -C)

# initial guess near power-law behavior
p0 = [15, 200, 0.3]
popt, _ = curve_fit(stretched, tr["days"].values, tr["log_price"].values,
                    p0=p0, maxfev=100000)
A, B, C = popt
se_pred_te = stretched(te["days"].values, *popt)
se_rmse = np.sqrt(np.mean((te["log_price"].values - se_pred_te)**2))

print(f"Plain power law:        log(P) = {a_pl:.2f} + {b_pl:.3f}*log(days)")
print(f"  OOS RMSE (2026): {pl_rmse:.4f}")
print(f"\nStretched exponential:  log(P) = {A:.3f} - {B:.3f}*days^(-{C:.4f})")
print(f"  OOS RMSE (2026): {se_rmse:.4f}")

# What each predicts for today vs actual
today_days = d["days"].iloc[-1]
actual_today = d["btc_price"].iloc[-1]
pl_today = np.exp(a_pl + b_pl*np.log(today_days))
se_today = np.exp(stretched(today_days, *popt))
print(f"\nActual price today:        ${actual_today:,.0f}")
print(f"Power law predicts:        ${pl_today:,.0f}  ({pl_today/actual_today:.2f}x)")
print(f"Stretched exp predicts:    ${se_today:,.0f}  ({se_today/actual_today:.2f}x)")

# Plot
plt.figure(figsize=(13,6))
plt.plot(d.index, d["log_price"], lw=0.6, color="gray", label="log price")
plt.plot(d.index, a_pl+b_pl*d["log_days"], "r-", lw=1.5, label="power law")
plt.plot(d.index, stretched(d["days"].values,*popt), "b-", lw=1.5, label="stretched exp")
plt.axvline(pd.Timestamp(CUTOFF), color="k", ls=":", label="train/test")
plt.legend(); plt.title("Power law vs stretched exponential")
plt.tight_layout(); plt.show()
```
