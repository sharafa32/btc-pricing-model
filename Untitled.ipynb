```python
import pandas as pd
import numpy as np

# Load the merged daily data
df = pd.read_csv("data/processed/factors_daily.csv", index_col=0, parse_dates=True)

print(f"Shape: {df.shape}")
print(f"Date range: {df.index.min().date()} to {df.index.max().date()}")
print(f"\nColumns: {list(df.columns)}")

# Check btc_price integrity — this is the critical column
print(f"\n--- btc_price checks ---")
print(f"  Zeros: {(df['btc_price'] == 0).sum()}")
print(f"  NaNs: {df['btc_price'].isna().sum()}")
print(f"  Negatives: {(df['btc_price'] < 0).sum()}")
print(f"  Min: {df['btc_price'].min()}")
print(f"  Max: {df['btc_price'].max()}")

# Check for date gaps (missing days)
full_range = pd.date_range(df.index.min(), df.index.max(), freq="D")
missing_days = len(full_range) - len(df.index.unique())
print(f"\n  Calendar days in range: {len(full_range)}")
print(f"  Actual rows: {len(df)}")
print(f"  Missing/duplicate days: {missing_days}")
```

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt
%matplotlib inline

GENESIS = pd.Timestamp("2009-01-03")
TRAIN_END = "2022-12-31"

# Days since genesis and logs
df["days_since_genesis"] = (df.index - GENESIS).days
df["log_price"] = np.log(df["btc_price"])
df["log_days"] = np.log(df["days_since_genesis"])

# Fit power-law trend on TRAINING DATA ONLY (leakage-free)
train = df.loc[:TRAIN_END]
X_train = sm.add_constant(train["log_days"])
y_train = train["log_price"]
trend_model = sm.OLS(y_train, X_train).fit()

alpha, beta = trend_model.params
print(f"Reference power-law trend (fit on 2015-2022 only):")
print(f"  alpha = {alpha:.4f}, beta = {beta:.4f}, R² = {trend_model.rsquared:.4f}")

# Project trend across ALL dates and compute deviation
df["trend"] = alpha + beta * df["log_days"]
df["deviation"] = df["log_price"] - df["trend"]

# Split deviation
dev_train = df.loc[:TRAIN_END, "deviation"]
dev_test = df.loc["2023-01-01":, "deviation"]

print(f"\nDeviation series:")
print(f"  Train ({len(dev_train)} obs): mean={dev_train.mean():.3f}, std={dev_train.std():.3f}")
print(f"  Test  ({len(dev_test)} obs): mean={dev_test.mean():.3f}, std={dev_test.std():.3f}")
print(f"  Full range: min={df['deviation'].min():.3f}, max={df['deviation'].max():.3f}")

# Sanity plot
fig, axes = plt.subplots(2, 1, figsize=(13, 7))
axes[0].plot(df.index, df["log_price"], label="log price", linewidth=0.8)
axes[0].plot(df.index, df["trend"], label="reference trend", color="red", linewidth=1.5)
axes[0].axvline(pd.Timestamp(TRAIN_END), color="gray", linestyle=":", label="train/test split")
axes[0].legend(); axes[0].set_title("Log price vs reference trend (fit on train only)")
axes[1].plot(df.index, df["deviation"], color="green", linewidth=0.8)
axes[1].axhline(0, color="black", linewidth=0.5)
axes[1].axvline(pd.Timestamp(TRAIN_END), color="gray", linestyle=":")
axes[1].set_title("Deviation (SETAR target)")
plt.tight_layout(); plt.show()
```

```python
# Load the FULL price history (independent of factor start date)
price = pd.read_csv("data/raw/btc_price.csv", index_col="date", parse_dates=True)

TREND_START = "2013-01-01"
TRAIN_END = "2022-12-31"

# Fit trend on 2013+ training data only
price_fit = price.loc[TREND_START:TRAIN_END].copy()
price_fit["days"] = (price_fit.index - GENESIS).days
price_fit["log_price"] = np.log(price_fit["btc_price"])
price_fit["log_days"] = np.log(price_fit["days"])

X = sm.add_constant(price_fit["log_days"])
trend_model = sm.OLS(price_fit["log_price"], X).fit()
alpha, beta = trend_model.params
print(f"Trend fit on {TREND_START} to {TRAIN_END}:")
print(f"  alpha={alpha:.4f}, beta={beta:.4f}, R²={trend_model.rsquared:.4f}")

# Now build the deviation on the FACTOR window (2015+, where df lives)
# so it aligns with everything else, but using the better-fit trend
df["days_since_genesis"] = (df.index - GENESIS).days
df["log_price"] = np.log(df["btc_price"])
df["log_days"] = np.log(df["days_since_genesis"])
df["trend"] = alpha + beta * df["log_days"]
df["deviation"] = df["log_price"] - df["trend"]

dev_train = df.loc[:TRAIN_END, "deviation"]
dev_test = df.loc["2023-01-01":, "deviation"]
print(f"\nDeviation (on 2015+ factor window, using 2013-fit trend):")
print(f"  Train: mean={dev_train.mean():.3f}, std={dev_train.std():.3f}")
print(f"  Test:  mean={dev_test.mean():.3f}, std={dev_test.std():.3f}")
print(f"  Range: [{df['deviation'].min():.3f}, {df['deviation'].max():.3f}]")
```

```python
from sklearn.metrics import mean_squared_error

# Prepare deviation series aligned with the factor window
dev = df["deviation"].dropna()
TRAIN_END = "2022-12-31"

dev_train = dev.loc[:TRAIN_END]
dev_test = dev.loc["2023-01-01":]

# Build lag-1 series (SETAR needs y_t and y_{t-1})
def build_lag_frame(series):
    lag_df = pd.DataFrame({
        "y": series,
        "y_lag1": series.shift(1)
    }).dropna()
    return lag_df

lag_train = build_lag_frame(dev_train)
lag_test = build_lag_frame(dev_test)

# --- Threshold search ---
# Grid over inner 60% of training-period deviation values
low_q, high_q = dev_train.quantile(0.20), dev_train.quantile(0.80)
threshold_grid = np.linspace(low_q, high_q, 100)

def fit_setar_at_threshold(lag_df, threshold):
    """Fit 2-regime AR(1) SETAR at a given threshold. Returns coeffs + SSR."""
    below = lag_df[lag_df["y_lag1"] <= threshold]
    above = lag_df[lag_df["y_lag1"] > threshold]
    if len(below) < 30 or len(above) < 30:
        return None
    # Regime 1: y = a1 + b1 * y_lag1
    X1 = sm.add_constant(below["y_lag1"])
    m1 = sm.OLS(below["y"], X1).fit()
    # Regime 2
    X2 = sm.add_constant(above["y_lag1"])
    m2 = sm.OLS(above["y"], X2).fit()
    ssr = m1.ssr + m2.ssr
    return {"threshold": threshold, "m1": m1, "m2": m2, "ssr": ssr,
            "n_below": len(below), "n_above": len(above)}

# Search
results = []
for t in threshold_grid:
    fit = fit_setar_at_threshold(lag_train, t)
    if fit:
        results.append(fit)

best = min(results, key=lambda r: r["ssr"])
print(f"Optimal threshold: {best['threshold']:.4f}")
print(f"  Below-regime obs: {best['n_below']} | Above-regime obs: {best['n_above']}")
print(f"  Regime 1 (below):  a={best['m1'].params[0]:.4f}, b={best['m1'].params[1]:.4f}")
print(f"  Regime 2 (above):  a={best['m2'].params[0]:.4f}, b={best['m2'].params[1]:.4f}")

# --- Out-of-sample prediction ---
def predict_setar(lag_df, best):
    """Apply the fitted SETAR to new data."""
    y_lag = lag_df["y_lag1"].values
    a1, b1 = best["m1"].params
    a2, b2 = best["m2"].params
    below_mask = y_lag <= best["threshold"]
    preds = np.where(below_mask, a1 + b1 * y_lag, a2 + b2 * y_lag)
    return preds

train_preds = predict_setar(lag_train, best)
test_preds = predict_setar(lag_test, best)

train_rmse = np.sqrt(mean_squared_error(lag_train["y"], train_preds))
test_rmse = np.sqrt(mean_squared_error(lag_test["y"], test_preds))

# Baseline: predict deviation = 0 (the "power law only" model)
baseline_test_rmse = np.sqrt(mean_squared_error(lag_test["y"], np.zeros_like(lag_test["y"])))

# Baseline 2: naive AR(1) with no threshold (single regime)
X_naive = sm.add_constant(lag_train["y_lag1"])
m_naive = sm.OLS(lag_train["y"], X_naive).fit()
naive_test_preds = m_naive.params[0] + m_naive.params[1] * lag_test["y_lag1"].values
naive_test_rmse = np.sqrt(mean_squared_error(lag_test["y"], naive_test_preds))

print(f"\n=== Results ===")
print(f"SETAR train RMSE:      {train_rmse:.4f}")
print(f"SETAR test RMSE:       {test_rmse:.4f}")
print(f"Naive AR(1) test RMSE: {naive_test_rmse:.4f}")
print(f"Power-law test RMSE:   {baseline_test_rmse:.4f}")
```

```python
HORIZON = 30

# Target: deviation H days ahead (what we're trying to predict)
df["target_fwd"] = df["deviation"].shift(-HORIZON)

# Build frame: predict future deviation from CURRENT deviation (threshold var = today's deviation)
fwd = pd.DataFrame({
    "y_future": df["target_fwd"],      # deviation 30 days from now
    "y_now": df["deviation"],           # today's deviation (threshold variable + predictor)
}).dropna()

fwd_train = fwd.loc[:TRAIN_END]
fwd_test = fwd.loc["2023-01-01":]

print(f"30-day-ahead SETAR")
print(f"  Train obs: {len(fwd_train)}, Test obs: {len(fwd_test)}")

# Threshold search on training data
low_q, high_q = fwd_train["y_now"].quantile(0.20), fwd_train["y_now"].quantile(0.80)
grid = np.linspace(low_q, high_q, 100)

def fit_at(frame, thr):
    below = frame[frame["y_now"] <= thr]
    above = frame[frame["y_now"] > thr]
    if len(below) < 30 or len(above) < 30:
        return None
    m1 = sm.OLS(below["y_future"], sm.add_constant(below["y_now"])).fit()
    m2 = sm.OLS(above["y_future"], sm.add_constant(above["y_now"])).fit()
    return {"thr": thr, "m1": m1, "m2": m2, "ssr": m1.ssr + m2.ssr,
            "n_below": len(below), "n_above": len(above)}

res = [fit_at(fwd_train, t) for t in grid]
res = [r for r in res if r]
best = min(res, key=lambda r: r["ssr"])

print(f"\nOptimal threshold: {best['thr']:.4f}")
print(f"  Below obs: {best['n_below']} | Above obs: {best['n_above']}")
print(f"  Regime 1 (below):  a={best['m1'].params.iloc[0]:.4f}, b={best['m1'].params.iloc[1]:.4f}")
print(f"  Regime 2 (above):  a={best['m2'].params.iloc[0]:.4f}, b={best['m2'].params.iloc[1]:.4f}")

def predict(frame, best):
    yn = frame["y_now"].values
    a1, b1 = best["m1"].params.iloc[0], best["m1"].params.iloc[1]
    a2, b2 = best["m2"].params.iloc[0], best["m2"].params.iloc[1]
    return np.where(yn <= best["thr"], a1 + b1*yn, a2 + b2*yn)

test_pred = predict(fwd_test, best)
test_rmse = np.sqrt(mean_squared_error(fwd_test["y_future"], test_pred))

# Naive AR (single regime, no threshold)
m_naive = sm.OLS(fwd_train["y_future"], sm.add_constant(fwd_train["y_now"])).fit()
naive_pred = m_naive.params.iloc[0] + m_naive.params.iloc[1] * fwd_test["y_now"].values
naive_rmse = np.sqrt(mean_squared_error(fwd_test["y_future"], naive_pred))

# Baseline: predict deviation reverts to 0
zero_rmse
```
